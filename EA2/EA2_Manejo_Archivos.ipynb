# Evidencia de Aprendizaje 2: Manejo de Archivos CSV y JSON
**Objetivo:** Importar, modificar y guardar archivos descargados manualmente desde Hugging Face utilizando clases propias y las librerías nativas `csv` y `json`.

```python
from manejador_csv import ManejadorCSV
from manejador_json import ManejadorJSON
from generador_resumen_txt import GeneradorResumenTXT

# URLs de origen específicas de los archivos descargados manualmente
URL_CSV = "https://huggingface.co/datasets/JasonZhouTI/Fast_Food_Restaurants/resolve/main/FastFoodRestaurants.csv"
URL_JSON = "https://huggingface.co/datasets/yahma/alpaca-cleaned/resolve/main/alpaca_data_cleaned.json"
```

### Procesamiento del CSV

```python
# 1. Instanciar clase e importar datos
gestor_csv = ManejadorCSV("datos/original.csv")
datos_csv = gestor_csv.importar_csv()

print("--- CSV ANTES DE MODIFICAR (Fila 1) ---")
print("Encabezados:", gestor_csv.encabezados)
print("Fila 1:", datos_csv[1])

# 2. Aplicar modificación (Cambiamos el nombre del restaurante en la primera fila de datos)
gestor_csv.actualizar_celda(fila_idx=1, nombre_columna="name", nuevo_valor="McDonald's Premium")

# 3. Guardar el nuevo archivo
ruta_nuevo_csv = gestor_csv.guardar_csv("datos/modificado.csv")

print("\n--- CSV DESPUÉS DE MODIFICAR (Fila 1) ---")
print("Fila 1:", gestor_csv.datos[1])
print(f"✅ Nuevo CSV guardado en: {ruta_nuevo_csv}")
```

### Procesamiento del JSON

```python
# 1. Instanciar clase e importar datos
gestor_json = ManejadorJSON("datos/test.json")
datos_json = gestor_json.importar_json()

print("--- JSON ANTES DE MODIFICAR (Registro 0) ---")
print(datos_json[0])

# 2. Aplicar modificación (Cambiamos el valor de la clave del primer registro)
clave_a_modificar = list(datos_json[0].keys())[0]
gestor_json.actualizar_registro(indice=0, clave=clave_a_modificar, nuevo_valor="MODIFICADO_PARA_AE2")

# 3. Guardar el nuevo archivo
ruta_nuevo_json = gestor_json.guardar_json("datos/modificado.json")

print("\n--- JSON DESPUÉS DE MODIFICAR (Registro 0) ---")
print(gestor_json.datos[0])
print(f"✅ Nuevo JSON guardado en: {ruta_nuevo_json}")
```

### Generación del Resumen TXT

```python
# Preparar los diccionarios con la información real de ejecución
info_csv = {
    "url_origen": URL_CSV,
    "archivo_entrada": str(gestor_csv.ruta_entrada),
    "archivo_salida": str(ruta_nuevo_csv),
    "clase": "ManejadorCSV (manejador_csv.py)",
    "metodos": ["importar_csv", "actualizar_celda", "guardar_csv"],
    "total_registros": len(gestor_csv.datos),
    "modificaciones": gestor_csv.modificaciones,
}

info_json = {
    "url_origen": URL_JSON,
    "archivo_entrada": str(gestor_json.ruta_entrada),
    "archivo_salida": str(ruta_nuevo_json),
    "clase": "ManejadorJSON (manejador_json.py)",
    "metodos": ["importar_json", "actualizar_registro", "guardar_json"],
    "total_registros": len(gestor_json.datos),
    "modificaciones": gestor_json.modificaciones,
}

# Generar y mostrar el resumen final
generador_txt = GeneradorResumenTXT("resumen.txt")
contenido_resumen = generador_txt.generar_resumen(info_csv, info_json)

print(contenido_resumen)
```
